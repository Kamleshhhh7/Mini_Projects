```python
import json
```

```python
#Load data
def load_data():
  with open("store_data.json","r") as f:
    sd=json.load(f)
    

    return sd
```

```python
data=load_data()
print(data)
```

```python
# Clean and structure data
def clean_data(data):
    text_to_num = {"one": 1, "two": 2, "three": 3, "four": 4, "five": 5}
    cleaned_data = []
    unique_users = set()

    # data consistency
    for user in data:
        raw_rating = str(user["rating"]).strip().lower()
        if raw_rating in text_to_num:
            raw_rating = text_to_num[raw_rating]
        user["rating"] = raw_rating

        # handling missing values
        raw_age = user.get("age")
        if raw_age is None:
            user["age"] = None

        # Deduplication
        name = user["name"].strip()
        if name in unique_users:
            continue

        unique_users.add(name)
        cleaned_data.append(user)

    return cleaned_data
```

```python
data=clean_data(data)
```

```python
#Getting meaningful insights:
def insight(data):
    total_rating=0
    poor_rating=0

    for user in data:
        total_rating += float(user["rating"])
    print(f"The average of rating is: {total_rating/len(data) }")

    for user in data:
        if(float(user["rating"]) < 3):
            poor_rating += 1
            print(f"The name of poor rater is{user["name"]}")
        
    print(f"Thepoor rating % is: {poor_rating/len(data) * 100}%")
        
    
```

```python
mf=insight(data)
```

```python
#recommndation (simple):
def recommendation(data):
    recommndate=[]

    for user in data:
        curr_recom={}
        curr_recom["name"]=user["name"]
        
        if(float(user["rating"])>=4):
            curr_recom["brand"] = "Apple"
        else:
            curr_recom["brand"] = "Samsung"

        recommndate.append(curr_recom)
    return recommndate
            
            
```

```python
recommendation(data)
```
